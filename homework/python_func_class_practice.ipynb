### Завдання 10. Перевірка простого числа ⭐⭐

Створіть функцію `is_prime(n)`, яка повертає `True`, якщо число просте, і `False` інакше.

**Нагадування:** просте число ділиться лише на 1 і на саме себе. Числа менші за 2 — не прості.

**Перевірте:**
```python
print(is_prime(7))    # True
print(is_prime(10))   # False
print(is_prime(1))    # False
```

> 💡 **Підказка:** У циклі від 2 до n-1 перевіряйте, чи ділиться `n` на якесь число без остачі. Якщо знайшли дільник — повертайте одразу `False`. Якщо цикл завершився без знайдених дільників — `True`.

````python
# Ваш код для завдання 10:
def is_prime(n):
  if n < 2:
    return False
  for i in range(2, n -1):
    if n % i == 0:
      return False
  return True

print(is_prime(7))
print(is_prime(10))
print(is_prime(1))
````

### Завдання 13. Середнє арифметичне ⭐⭐

Створіть функцію `average(numbers)`, яка приймає список чисел і повертає їхнє середнє арифметичне. Якщо список порожній — поверніть 0.

> 💡 **Підказка:** Не забудьте перевірити порожній список перед діленням, щоб уникнути ділення на нуль.

````python
# Ваш код для завдання 13:
def average(numbers):
  if len(numbers) == 0:
    return 0

  total = 0

  for number in numbers:
    total += number
  return total / len(numbers)

print(average([7,5,9,2,65]))
````

### Завдання 14. Підрахунок цифр у числі ⭐⭐⭐

Створіть функцію `count_digits(n)`, яка приймає ціле число і повертає кількість цифр у ньому.

**Перевірте:**
```python
print(count_digits(12345))   # 5
print(count_digits(7))       # 1
print(count_digits(0))       # 1
```

Реалізуйте через **математику** (без перетворення в рядок).

> 💡 **Підказка:** У циклі `while n > 0` ділимо число націло на 10 (`n = n // 10`) і збільшуємо лічильник. Окремо обробіть випадок `n == 0`.

````python
# Ваш код для завдання 14:
def count_digits(n):
  if n == 0:
    return 1

  cnt = 0
  while n > 0:
    n = n // 10
    cnt += 1
  return cnt
print(count_digits(1255))
````

### Завдання 16. Найдовше слово ⭐⭐⭐

Створіть функцію `longest_word(sentence)`, яка приймає рядок-речення і повертає найдовше слово.

**Перевірте:**
```python
print(longest_word("Python це чудова мова програмування"))   # "програмування"
```

> 💡 **Підказка:** Розбийте речення на список слів через `.split()`. Потім переберіть і знайдіть найдовше, порівнюючи через `len()`.

````python
# Ваш код для завдання 16:
def longest_word(sentence):
  words = sentence.split()
  max_word = ""

  for word in words:
    if len(word) > len(max_word):
      max_word = word
  return max_word

print(longest_word("Найпростіший спосіб: повернути результат порівняння"))
````

### Завдання 17. Калькулятор з операцією ⭐⭐⭐

Створіть функцію `calculate(a, b, operation)`, яка виконує одну з операцій залежно від рядка `operation`:
- `"+"` — додавання
- `"-"` — віднімання
- `"*"` — множення
- `"/"` — ділення

Якщо операція невідома — поверніть рядок `"Невідома операція"`. Якщо ділення на нуль — поверніть `"Помилка: ділення на нуль"`.

````python
# Ваш код для завдання 17:
def calculate(a, b, operation):
  if operation == "+":
        result = a + b
  elif operation == "-":
        result = a - b
  elif operation == "*":
        result = a * b
  elif operation == "/":
      if b != 0:
            result = a / b
      else:
            result = "Error: division by 0"
  else:
        result = "Unknown operation"
  return result

print(calculate(1894, 77, "/"))
````

### Завдання 18. Числа Фібоначчі ⭐⭐⭐

Створіть функцію `fibonacci(n)`, яка повертає **список** з перших `n` чисел Фібоначчі.

**Нагадування:** кожне число — сума двох попередніх. Перші два: 0 і 1.

**Перевірте:**
```python
print(fibonacci(7))   # [0, 1, 1, 2, 3, 5, 8]
```

> 💡 **Підказка:** Створіть список з перших двох чисел `[0, 1]`. У циклі додавайте суму двох останніх елементів через `.append()`.

````python
# Ваш код для завдання 18:
def fibonacci(n):
  fib = [0, 1]

  for i in range(2, n):
    next_num = fib[-1] + fib[-2]
    fib.append(next_num)
  return fib[:n]

print(fibonacci(7))
````

````python
# Ваш код для завдання 22:
class Circle:
  def __init__(self, radius):
    self.radius = radius

  def area(self):
    area = 3.14 * (self.radius ** 2)
    return area

  def circumference(self):
    cir = 2 * 3.14 * self.radius
    return cir

my_circle = Circle(5)
print(my_circle.area(), my_circle.circumference())
````

### Завдання 23. Клас Student ⭐⭐

Створіть клас `Student` з атрибутами `name` і порожнім списком `grades`. Додайте методи:
1. `add_grade(grade)` — додає оцінку до списку
2. `average()` — повертає середній бал (0, якщо оцінок ще немає)
3. `__str__()` — повертає рядок виду `"Студент <name>, середній бал: <average>"`

Створіть студента, додайте 4 оцінки, виведіть об'єкт через `print()`.

````python
# Ваш код для завдання 23:
class Student:
  def __init__(self, name):
    self.name = name
    self.grades = []

  def add_grade(self, grade):
    self.grades.append(grade)

  def average(self):
    if len(self.grades) == 0:
      return 0
    return sum(self.grades) / len(self.grades)

  def __str__(self):
    return f"Студент {self.name}, середній бал: {self.average()}"

stud_1 = Student("Max")
stud_1.add_grade(95)
stud_1.add_grade(80)
stud_1.add_grade(85)
stud_1.add_grade(90)

print(stud_1)

````

### Завдання 24. Клас Counter ⭐⭐

Створіть клас `Counter` (лічильник). У конструкторі задайте `self.value = 0`. Додайте методи:
1. `increment()` — збільшити на 1
2. `decrement()` — зменшити на 1
3. `reset()` — скинути в 0
4. `__str__()` — повернути поточне значення як рядок

Створіть лічильник, виконайте 5 increment і 2 decrement, виведіть результат.

````python
# Ваш код для завдання 24:
class Counter:
  def __init__(self):
    self.value = 0

  def increment(self):
    self.value += 1

  def decrement(self):
    self.value -= 1

  def reset(self):
    self.value = 0

  def __str__(self):
    return str(self.value)

cnt = Counter()
for i in range(5):
  cnt.increment()

for i in range(2):
  cnt.decrement()

print(cnt)
````

Створіть клас `Team` з полями `name`, `wins`, `draws`, `losses`. Додайте метод `points()` — повертає очки команди (3 за перемогу, 1 за нічию).

Створіть список з 6 команд з різними результатами.

Напишіть **функції**:
1. `tournament_table(teams)` — виводить турнірну таблицю, відсортовану за очками (від найбільших)
2. `champion(teams)` — повертає команду з найбільшою кількістю очок
3. `total_matches(teams)` — загальна кількість зіграних матчів (сума всіх перемог + нічиїх + поразок, поділена на 2, бо кожен матч рахується двічі)

````python
# Ваш код для завдання 36:
class Team:
    def __init__(self, name, wins, draws, losses):
        self.name = name
        self.wins = wins
        self.draws = draws
        self.losses = losses

    def points(self):
        return self.wins * 3 + self.draws

def tournament_table(teams):
    sorted_teams = sorted(teams, key=lambda t: t.points(), reverse=True)

    print("Турнірна таблиця:")

    for team in sorted_teams:
        print(f"{team.name}: {team.points()} очок")

def champion(teams):
    return max(teams, key=lambda t: t.points())

def total_matches(teams):
    total = sum(team.wins + team.draws + team.losses for team in teams)
    return total // 2

teams = [
    Team("Team A", 10, 2, 3),
    Team("Team B", 8, 5, 2),
    Team("Team C", 7, 4, 4),
    Team("Team D", 6, 3, 6),
    Team("Team E", 5, 5, 5),
    Team("Team F", 3, 2, 10)
]

tournament_table(teams)

best = champion(teams)
print("\nЧемпіон:", best.name)

print("Загальна кількість матчів:", total_matches(teams))
````

Створіть клас `BankAccount` з полями `owner`, `balance`, методами `deposit(amount)`, `withdraw(amount)`, `__str__()`.

Створіть **функції** для роботи з кількома рахунками:
1. `transfer(from_account, to_account, amount)` — переказує гроші з одного рахунку на інший (перевірте, чи вистачає коштів)
2. `total_money(accounts)` — загальна сума на всіх рахунках
3. `richest_client(accounts)` — рахунок з найбільшим балансом
4. `accounts_above(accounts, min_balance)` — список рахунків з балансом > `min_balance`

Створіть 4-5 рахунків і протестуйте все.

> 💡 **Підказка:** У `transfer` спочатку зробіть `withdraw` з першого рахунку. Якщо вийшло (баланс зменшився) — зробіть `deposit` на другий. Можете додати метод `withdraw()`, який повертає `True`/`False` залежно від успіху.

````python
# Ваш код для завдання 37:
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            return True
        return False

    def __str__(self):
        return f"{self.owner}: {self.balance} грн"

def transfer(from_account, to_account, amount):
    if from_account.withdraw(amount):
        to_account.deposit(amount)
        print("Переказ успішний")
    else:
        print("Недостатньо коштів")

def total_money(accounts):
    return sum(account.balance for account in accounts)

def richest_client(accounts):
    return max(accounts, key=lambda a: a.balance)

def accounts_above(accounts, min_balance):
    return [account.owner for account in accounts if account.balance > min_balance]

a1 = BankAccount("Anna", 5000)
a2 = BankAccount("Oleg", 3000)
a3 = BankAccount("Maria", 8000)
a4 = BankAccount("Ivan", 2000)

accounts = [a1, a2, a3, a4]

transfer(a1, a2, 1000)

print("\nЗагальна сума:", total_money(accounts))

richest = richest_client(accounts)
print("Найбагатший клієнт:", richest.owner)

print("Баланс більше 4000:")
print(accounts_above(accounts, 4000))

print("\nРахунки:")
for account in accounts:
    print(account)
````

````python
# Ваш код для завдання 38:
class Student:
    def __init__(self, name):
        self.name = name
        self.grades = []

    def add_grade(self, grade):
        self.grades.append(grade)

    def average(self):
        if len(self.grades) == 0:
            return 0
        return sum(self.grades) / len(self.grades)


class SchoolClass:
    def __init__(self, name):
        self.name = name
        self.students = []

    def add_student(self, student):
        self.students.append(student)

    def class_average(self):
        total = sum(student.average() for student in self.students)
        return total / len(self.students)

def best_class(classes):
    return max(classes, key=lambda c: c.class_average())

def total_students(classes):
    return sum(len(c.students) for c in classes)

def school_report(classes):
    print("Звіт по школі:\n")

    for c in classes:
        print(f"{c.name}:")
        print(f"Кількість учнів: {len(c.students)}")
        print(f"Середній бал: {c.class_average():.2f}\n")

c1 = SchoolClass("7-A")

s1 = Student("Anna")
s1.add_grade(10)
s1.add_grade(12)

s2 = Student("Oleg")
s2.add_grade(8)
s2.add_grade(9)

c1.add_student(s1)
c1.add_student(s2)

c2 = SchoolClass("7-B")

s3 = Student("Maria")
s3.add_grade(11)
s3.add_grade(12)

s4 = Student("Ivan")
s4.add_grade(7)
s4.add_grade(8)

c2.add_student(s3)
c2.add_student(s4)

c3 = SchoolClass("7-C")

s5 = Student("Kate")
s5.add_grade(9)
s5.add_grade(10)

s6 = Student("Max")
s6.add_grade(6)
s6.add_grade(7)

c3.add_student(s5)
c3.add_student(s6)


classes = [c1, c2, c3]

best = best_class(classes)

print("Найкращий клас:", best.name)

print("Загальна кількість учнів:", total_students(classes))

print()
school_report(classes)
````

### Завдання 39. Менеджер задач (todo) ⭐⭐⭐

Створіть клас `Task` з полями `title`, `priority` (1-3, де 1 — найвищий), `done = False`. Додайте метод `mark_done()`.

Створіть **функції**, які приймають список задач:
1. `add_task(tasks, title, priority)` — додає нову задачу в список
2. `pending_tasks(tasks)` — список невиконаних задач
3. `tasks_by_priority(tasks, priority)` — задачі певного пріоритету
4. `summary(tasks)` — виводить статистику: загалом, виконано, залишилось, найвищий пріоритет серед невиконаних

````python
# Ваш код для завдання 39:
class Task:
    def __init__(self, title, priority):
        self.title = title
        self.priority = priority
        self.done = False

    def mark_done(self):
        self.done = True

def add_task(tasks, title, priority):
    tasks.append(Task(title, priority))

def pending_tasks(tasks):
    return [task.title for task in tasks if not task.done]

def tasks_by_priority(tasks, priority):
    return [task.title for task in tasks if task.priority == priority]

def summary(tasks):
    total = len(tasks)
    completed = len([task for task in tasks if task.done])
    pending = total - completed

    not_done = [task.priority for task in tasks if not task.done]

    if len(not_done) > 0:
        highest_priority = min(not_done)
    else:
        highest_priority = None

    print("Статистика:")
    print("Всього задач:", total)
    print("Виконано:", completed)
    print("Залишилось:", pending)
    print("Найвищий пріоритет серед невиконаних:", highest_priority)

tasks = []

add_task(tasks, "Homework", 1)
add_task(tasks, "Clean room", 3)
add_task(tasks, "Buy food", 2)

tasks[1].mark_done()

print("Невиконані задачі:")
print(pending_tasks(tasks))

print("\nПріоритет 1:")
print(tasks_by_priority(tasks, 1))

print()
summary(tasks)
````

### Завдання 40. Аналіз ресторанних замовлень 🔥 ⭐⭐⭐

Велике завершальне завдання, де всі попередні теми поєднуються разом.

**Створіть класи:**
1. `MenuItem` — поля `name`, `price`, `category`
2. `Order` — поля `table_number`, `items` (список `MenuItem`), метод `total()` (сума цін), метод `add_item(item)`
3. `Restaurant` — поле `orders` (список замовлень), метод `add_order(order)`, метод `total_revenue()` (сумарна виручка), метод `most_popular_category()`

**Створіть функції**, які приймають об'єкт `Restaurant`:
1. `busiest_table(restaurant)` — стіл з найбільшою сумою замовлень
2. `category_breakdown(restaurant)` — словник `{категорія: виручка}`
3. `print_summary(restaurant)` — повний звіт по ресторану

**Створіть тестові дані:**
- 6-8 страв у меню
- 4-5 замовлень з різних столиків (по 3-4 страви у кожному)

Викличте всі функції і виведіть результати.

> 💡 **Підказка:** Це велике завдання — виконуйте поступово. Спочатку створіть і протестуйте `MenuItem`. Потім `Order` — додавайте страви, перевіряйте `total()`. Потім `Restaurant`. Тільки після цього пишіть функції.

````python
# Ваш код для завдання 40:
class MenuItem:
    def __init__(self, name, price, category):
        self.name = name
        self.price = price
        self.category = category

class Order:
    def __init__(self, table_number):
        self.table_number = table_number
        self.items = []

    def add_item(self, item):
        self.items.append(item)

    def total(self):
        return sum(item.price for item in self.items)

class Restaurant:
    def __init__(self):
        self.orders = []

    def add_order(self, order):
        self.orders.append(order)

    def total_revenue(self):
        return sum(order.total() for order in self.orders)

    def most_popular_category(self):
        categories = {}

        for order in self.orders:
            for item in order.items:
                if item.category in categories:
                    categories[item.category] += 1
                else:
                    categories[item.category] = 1

        return max(categories, key=categories.get)

def busiest_table(restaurant):
    return max(restaurant.orders, key=lambda o: o.total())

def category_breakdown(restaurant):
    result = {}

    for order in restaurant.orders:
        for item in order.items:
            if item.category in result:
                result[item.category] += item.price
            else:
                result[item.category] = item.price

    return result

def print_summary(restaurant):
    print("ЗВІТ ПО РЕСТОРАНУ\n")

    for order in restaurant.orders:
        print(f"Стіл {order.table_number}: {order.total()} грн")

    print("\nЗагальна виручка:", restaurant.total_revenue())

    print("Найпопулярніша категорія:",
          restaurant.most_popular_category())

    print("\nВиручка по категоріях:")
    print(category_breakdown(restaurant))

pizza = MenuItem("Pizza", 250, "Main")
burger = MenuItem("Burger", 180, "Main")
cola = MenuItem("Cola", 50, "Drink")
coffee = MenuItem("Coffee", 70, "Drink")
cake = MenuItem("Cake", 120, "Dessert")
salad = MenuItem("Salad", 150, "Starter")

o1 = Order(1)
o1.add_item(pizza)
o1.add_item(cola)
o1.add_item(cake)

o2 = Order(2)
o2.add_item(burger)
o2.add_item(coffee)

o3 = Order(3)
o3.add_item(pizza)
o3.add_item(salad)
o3.add_item(cola)

o4 = Order(4)
o4.add_item(cake)
o4.add_item(coffee)
o4.add_item(cola)

restaurant = Restaurant()

restaurant.add_order(o1)
restaurant.add_order(o2)
restaurant.add_order(o3)
restaurant.add_order(o4)

best_table = busiest_table(restaurant)

print("Стіл з найбільшою сумою:", best_table.table_number)

print("\n")
print_summary(restaurant)
````
