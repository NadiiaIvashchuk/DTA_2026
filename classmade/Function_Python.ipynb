# **Type of function results**

## NoneType
NoneType - it is type of data

````python
my_dict = {
    1: "one",
    2: "two",
    3: "three",
    4: "four",
    5: "five"
}
````

````python
# NonType, without chage data and output result:
def info(d):
    print(d)

#info(my_dict)
type(info(my_dict))
````

````python
# NonType, with chage data and output result:
def add_el_view(d):
    d[6] = "six"
    print(d)

#info(my_dict)
type(add_el_view(my_dict))
````

````python
# NonType, with chage data witout output result:
def add_el(d):
    d[6] = "six"

print(type(add_el(my_dict)))
print(f"my dict: {my_dict}")
````

## The function return a result for processing

````python
def sum_num(a, b):
    return a + b

print(type(sum_num(5, 7)))
print(sum_num(5, 7))

print()

sum_num(2, 8) + 10
````

````python
my_dict = {
    1: "one",
    2: "two",
    3: "three",
    4: "four",
    5: "five"
}
````

````python
def dict_value(d, key):
    print(d[key])  # my_dict[3]

print(type(dict_value(my_dict, 3)))
dict_value(my_dict, 3)
````

````python
def dict_value(d, key):
    return d[key]  # my_dict[3]

dict_value(my_dict, 3) + " 333"
````

````python
def new_func(d : dict, key: int) -> str:
    return d[key]

new_func("qwerty", 3) + " 333"
````
